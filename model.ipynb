```python
from ultralytics import YOLO
```

```python
model = YOLO("yolov8n.pt")  # или "yolov8s.pt", "yolov8m.pt", "yolov8l.pt", "yolov8x.pt"
```

```python
results = model("bear_2.jpg")  # путь к файлу

# Визуализировать результаты
results[0].show()  # отобразит окно с bbox
results[0].save(filename="output.jpg")  # сохранит результат
```

```python
results = model("vidosik.mp4", stream=True)

for r in results:
    r.show()
```
